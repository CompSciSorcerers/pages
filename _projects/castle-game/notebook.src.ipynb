# Castle Adventure Game

Embark on an epic journey through a medieval castle filled with challenges and obstacles. Each level presents unique gameplay mechanics and increasingly difficult challenges.

## Game Levels

Complete the levels in order to progress through the castle:

1. **Outside** - Navigate the castle exterior and learn basic movement
2. **Maze** - Find your way through the castle maze
3. **Archery** - Test your skills with bow and arrow
4. **Fortress** - Face the ultimate challenge in the castle fortress

## Level Details

### Level 1: Outside
**GameLevelOutside** - Learn the basic controls and navigate the castle exterior. Get familiar with movement mechanics and explore the surroundings.

### Level 2: Maze
**GameLevelMaze** - Navigate through the intricate castle maze. Use your wits to find the correct path and avoid dead ends.

### Level 3: Archery
**GameLevelArchery** - Test your precision and timing in the archery range. Hit moving targets and score points to advance.

### Level 4: Fortress
**GameLevelFortress** - Face the ultimate challenge with enemy scythes, boss battles, and special abilities. Use the Space key to fire interceptors and defend yourself!
